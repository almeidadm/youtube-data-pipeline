```python

from youtube_data_pipeline import YouTubeDataClient, YouTubeDataConfig
```

```python
config = YouTubeDataConfig(
    api_key="YOUTUBE_API_KEY",
    storage_root="./data/raw",
)

client = YouTubeDataClient.from_config(config)
```

```python
result = client.fetch(
    endpoint="search",
    params={
        "part": "snippet",
        "q": "imposto de 44% sobre aluguel por temporada",
        "maxResults": 25,
        "type": "video",
        "regionCode": "BR",
    },
    max_pages=20,
)

print(result["files"])
print(result["credit_state"])
```

```python
import json 

with open(result["files"][0], "r") as f:
    data = json.load(f)
print(json.dumps(data, indent=2))
```

```python
videos = [item for item in data["items"] if item["id"]["kind"] == "youtube#video"]
```

```python

video_detail = client.fetch(
    endpoint="videos",
    params={
        "part": "snippet,statistics,recordingDetails,contentDetails",
        "id": ",".join([video["id"]["videoId"] for video in videos])
    }
)
print(video_detail["credit_state"])
```

```python
with open(video_detail["files"][0], "r") as f:
    data = json.load(f)
print(json.dumps(data, indent=2))
```

```python
comments_threads = client.fetch(
    endpoint="commentThreads",
    params={
        "part": "snippet,replies",
        "videoId": videos[1]["id"]["videoId"],
    },
    max_pages=50,
)
```

```python
with open(comments_threads["files"][0], "r") as f:
    data = json.load(f)
print(json.dumps(data, indent=2))
```

```python
client.fetch(
    endpoint="comments",
    params={
        "part": "snippet",
        "id": data["items"][0]["id"]
    },
    max_pages=50,
)
```
